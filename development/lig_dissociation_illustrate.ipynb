```python
from architector import (build_complex,
                         view_structures) 
# The major routines used in architector
from architector import CalcExecutor, convert_io_molecule
```

```python
inputDict = {'core':{'metal':'Fe','coreCN':6}, #Specify metal coordination number (CN)
            'ligands':['water']+['terpy']+['bipy'], # Specify what is filling the coordination environment
            'parameters':{'assemble_method':'UFF',
                          'full_method':'UFF'}} # No additional parameters needed for default
fe_out = build_complex(inputDict) # Now just build using the dictionary!
```

```python
view_structures(fe_out)
```

```python
mol = convert_io_molecule(
    fe_out[list(fe_out.keys())[0]]['mol2string']
    ) # Pull out first structure
```

```python
trajs = mol.lig_dissociation_sample(
    max_dist=3, # Distance to push ligand away.
    steps=10, # Number of steps on trajectory
)
```

```python
view_structures(trajs[0][0:8]) # First Ligand trajectory
```

```python
view_structures(trajs[1][0:8]) # Second Ligand trajectory
```
