```python
from architector import build_complex, convert_io_molecule
from architector import view_structures
```

```python
out = build_complex({'core':{'metal':'Fe','coreCN':4},
                     'ligands':[],
                     'parameters':{'xtb_relax':False}})
```

```python
view_structures(out)
```

```python
mol = convert_io_molecule(out[list(out.keys())[1]]['mol2string'])
```

```python
view_structures(mol)
```

```python
asas = mol.get_asa(probe_radius=1.4,return_patch_midpoints=True)
asas
```

```python
ats = mol.ase_atoms.copy()
from ase import Atom
for thing in asas[2]:
    ats += Atom('La',position=thing['patch_midpoint'])
view_structures(ats)
```
