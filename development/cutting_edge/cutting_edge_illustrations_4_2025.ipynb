```python
from architector.io_arch_dock import add_non_covbound_species, defaults
from architector import build_complex, view_structures, convert_io_molecule
from architector.io_ptable import ligands_dict
```

```python
ligand = {'smiles':"[O-][C@H](Cl)C[C@@H](Br)[O-]",
          'coordList':[0,6],
          'ligType':'bi_cis'}
out = build_complex({'core':{'metal':'Fe','coreCN':6},
                     'ligands':[ligand]*3,
                     'parameters':{
                         'assemble_method':'UFF',
                         'full_method':'UFF',
                         'enforce_ligand_internal_symmetry':True,
                        'return_only_1':True}})
view_structures(out)
```

```python
# Make prettier
view_structures(
    out,
    labels=['chiral_fe1'],
    render_pymol=True,
    pymol_light_count=3,
    pymol_metal_scale=0.8,
    pymol_reflect_metal=100,
    pymol_shiny_metal=100,
)
```

```python
out = build_complex({'core':{'metal':'U','coreCN':6},
                     'ligands':['oxo']*2+['bipy'] + ['hydroxyl']*2,
                     'parameters':{
                         'metal_ox':6,
                         'assemble_method':'UFF',
                        'return_only_1':True}
                        })
view_structures(
    out,
    labels=['u_oxo1'],
    render_pymol=True,
    pymol_light_count=3,
    pymol_metal_scale=0.8,
    pymol_reflect_metal=100,
    pymol_shiny_metal=100,
)
mol = out[list(out.keys())[0]]['mol2string']
```

```python
view_structures(mol,labelinds=True)
```

```python
edit_mol = convert_io_molecule(mol)
edit_mol.functionalize_3D?
```

```python
edit_mol = convert_io_molecule(mol)
edit_mol.functionalize_3D(
    functional_groups=[mol],
    functionalization_inds=[[0, 0]], # "functional group" added to metal center
    functional_group_mol_inds=[[21, 23]], # Oxygens added briding to metal center
    remove_inds=[21,22,23,24], # Remove hydroxyls from first molecule
    remove_hydrogens_when_adding=[[False, False]],
    bond_orders = [[1,1]],
    uff_opt=True,
    xtb_opt=True
)
view_structures(edit_mol)
```

```python
view_structures(
    edit_mol,
    labels=['u_oxo2'],
    render_pymol=True,
    pymol_light_count=3,
    pymol_metal_scale=0.8,
    pymol_reflect_metal=100,
    pymol_shiny_metal=100,
)
```
